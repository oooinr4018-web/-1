# 1.1. 인공지능, 머신 러닝과 딥러닝

인공지능(AI): 인간의 지능을 모방하여 사람이 하는 일을 컴퓨터(기계)가 할 수 있도록 하는 기술

- 머신 러닝: 인공 신경망 서포트 벡터 머신 결정 트리

: 입력 데이터에 알고리즘을 적용하여 예측 수행

: 주어진 데이터를 인간이 먼저 전처리 (데이터 특징을 스스로 추출 X)

: 각 데이터 특성을 컴퓨터에 인식시키고 학습시켜 문제 해결

: 입력 -> 특성 추출 (인간의 처리) -> 분류 (컴퓨터의 처리) -> 출력 (일반적으로 숫자 값)

: 동일한 유형의 데이터 분석을 위한 재사용 불가

: 수천 개의 데이터 필요

: 훈련 기간: 단시간


- 딥러닝: CNN, RNN, RBM

: 정보를 전달하는 신경망을 사용하여 데이터 특징 및 관계 해석

: 대량의 데이터를 신경망에 적용 시 컴퓨터가 스스로 분석 후 답 추출

: 입력 -> 특성 추출 + 분류 (컴퓨터의 처리) -> 출력 (점수, 텍스트, 소리 등 어떤 것이든 가능)

: 동일한 유형의 데이터 분석을 위한 재사용 가능

: 수백만 개 이상의 데이터 필요

: 훈련 시간: 장시간

:

인공지능>머신 러닝> 딥러닝


# 1.2. 머신 러닝이란

머신 러닝: 인공지능의 한 분야로, 컴퓨터 스스로 대용량 데이터에서 지식, 패턴을 찾아 학습과 예측 수행

(컴퓨터 학습 알고리즘, 기술 개발 분야)




# 1.2.1. 머신 러닝 학습 과정

1. 머신 러닝 학습 과정

- 학습 단계

레이블 -> 분류/예측 모델 (모형)

데이터 (텍스트, 이미지) -> 특성 추출 -> 머신 러닝 알고리즘 적용 -> 분류/예측 모델 (모형)

- 예측 단계

데이터 (텍스트, 이미지) -> 특성 추출 -> 분류/예측 모델 (모형) (*학습 단계에서 진화 학습한 모형) -> 결괏값

- 특성 추출: 사람이 인지하는 데이터를 컴퓨터가 인지할 수 있는 데이터로 변환해주는 과정에서, 데이터별로 어떤 특징을 가지고 있는지 찾아내고, 이를 토대로 데이터를 벡터로 변환하는 작업

2. 머신 러닝의 주요 구성 요소

- 데이터: 머신 러닝이 학습 모델을 만드는 데 사용하는 것

: 실제 데이터의 특징을 잘 반영하고, 편향되지 않는 훈련 데이터 확보가 중요함.

: '훈련 데이터셋', '테스트 데이터셋'으로 분리해서 사용

(모델 성능 평가: '훈련 데이터셋'은 또 다시 '훈련 데이터셋'과 '검증 데이터셋'으로 분리 사용하기도 함. 검증 데이터셋은 훈련 데이터셋의 일부를 떼어서 사용하므로 좋지 않음.

하지만 검증 데이터셋을 통해 새롭게 수집될 데이터 예측 평가 가능 + 훈련 데이터셋에 대한 정확도는 높은데 검증 데이터셋에 대한 정확도가 낮은 경우, 훈련 데이터셋의 과적합 가능성(-> 에포크(epoch) or 정규화(normalization))

- 모델: 머신 러닝의 학슴 단계에서 얻은 최종 결과물 (=가설)

: 모델 (=가설) 선택 -> 모델 학습 및 평가 -> 평가 바탕 업그레이드

# 2.1.1. 파이토치 특징 및 장점

파이토치 특징: GPU에서 텐서 조작 및 동적 신경망 구축이 가능한 프레임워크

: 단순, 낮은 CPU 활용, 직관적인 인터페이스

- GPU

: 연산 속도를 빠르게 하는 역할

: API - CUDA, cuDNN

: 병렬 연산에서의 GPU 속도가 CPU보다 훨씬 빠름.

- 텐서

: 파이토치의 데이터 형태

: 단일 데이터 형식으로 된 자료들의 다차원 행렬

: 명령어(변수.cuda())로 GPU 연산 수행을 시킴.

: 3차원 이상의 배열 행렬 (행렬의 다차원 표현)

: torch.tensor()

- 동적 신경망

: 훈련을 반복할 때마다 네트워크 변경이 가능한 신경망

: 'Define by Run' 방식 - 연산 그래프 정의, 값 초기화 (분리해서 보지 X)

*연산 그래프: 노드(변수), 엣지(연산 담당)로 구성.

: 손실 함수 기울기 (가중치, 바이어스 기반 계산) -> 가중치 업데이트 (경사 하강법)


# 2.1.2. 파이토치의 아키텍처

- 파이토치 API

: 텐서 신경망 구축, 훈련 도움.

# 2.2. 파이토치 기초 문법

# 2.2.1. 텐서 다루기

1. 텐서 생성 및 변환

텐서 파이토치의 가장 기본이 되는 데이터 구조

```python
# 텐서 생성 코드
import torch
print(torch.tensor([[1,2],[3,4]])) # 2차원 형태의 텐서 생성
print(torch.tensor([[1,2],[3,4]], device="cuda:0")) # GPU에 텐서 생성
print(torch.tensor([[1,2],[3,4]], dtype=torch.float64)) # dtype을 이용하여 텐서 생성
```

```python
temp=torch.tensor([[1,2],[3,4]])
print(temp.numpy()) # 텐서를 ndarray로 변환

temp=torch.tensor([[1,2],[3,4]], device="cuda:0")
print(temp.to("cpu").numpy()) # GPU상의 텐서를 CPU의 텐서로 변환한 후, ndarray로 변환
```

# Sequential 신경망을 정의하는 방법

nn.Sequential: __init__()에서 사용할 모델 정의, forward() 함수에서 모델 속 계산 좀 더 가독성이 뛰어나게 코드 작성

: 모델 계층이 복잡할수록 효과가 뛰어남.

```python
import torch.nn as nn
class MLP(nn.Module):
  def __init__(self):
    super(MLP, self).__init__()
    self.layer1=nn.Sequential(
        nn.Conv2d(in_channels=3, out_channels=64, kernel_size=5),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2))

    self.layer2=nn.Sequential(
        nn.Conv2d(in_channels=64, out_channels=30, kernel_size=5),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(2))

    self.layer3=nn.Sequential(
        nn.Linear(in_features=30*5*5, out_features=10, bias=True),
        nn.ReLU(inplace=True))

    def forward(self, x):
      x=self.lyaer1(x)
      x=self.layer2(x)
      x=x.view(x.shape[0], -1)
      x=self.layer3(x)
      return x

model=MLP() #모델에 대한 객체 생성

print("Printing children\n-------------------------------------")
print(list(model.children()))
print("\n\nPrinting Modules\n-------------------------------------")
print(list(model.modules()))
```

# 함수로 신경망을 정의하는 밭법

모델이 복잡해짐. -> 함수 이용보다 nn.Module() 사용이 더 좋음.

```python
def MLP(in_features=1, hidden_features=20, out_features=1):
  hidden=nn.Linear(in_features=in_features, out_features=hidden_features,
                   bias=True)
  activation=nn.ReLU()
  output=nn.Linear(in_features=hidden_features, out_features=out_features,
                   bias=True)
  net=nn.Sequential(hidden, activation, output)
  return net
```
