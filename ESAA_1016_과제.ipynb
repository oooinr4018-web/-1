# 4.2.1. 딥러닝 용어

- 딥러닝은 입력층, 출력층, 두 개 이상의 은닉층으로 구성되어 있음.

입력층: 데이터를 받아들이는 층

은닉층: 모든 입력 노드부터 입력 값을 받아 가중합을 계산하고, 이 값을 활성화 함수에 적용하여 출력층에 전달하는 층

출력층: 신경망의 최종 결괏값이 포함된 층

가중치: 노드, 노드 간 연결 강도

: 입력 값이 연산 결과에 미치는 영향력을 조절하는 요소

바이어스: 가중합에 더해 주는 상수로, 하나의 뉴런에서 활성화 함수를 거쳐 최종적으로 출력되는 값을 조정하는 역할

가중합(전달 함수): 가중치와 신호의 곱을 합한 것

가중합 = 시그마(wi*xi+b)

: 각 노드에서 들어오는 신호에 가중치를 곱햇 다음 노드로 전달하는데, 이 값들을 모두 더한 합계

활성화 함수: 신호를 입력받아 이를 적절히 처리하여 출력해 주는 함수

: 전달 함수에서 전달받은 값을 출력할 때 일정 기준에 따라 출력 값을 변화시키는 비선형 함수

: 시그모이드, 하이포볼릭 탄젠트, 렐루 함수 등

- 시그모이드 함수

: 선형 함수의 결과를 0~1 사이에서 비선형 형태로 변형

: 결괏값의 평균이 0이 아닌 양수로 편향된 문제 해결

: 로지스틱 회귀와 같은 분류 문제를 확률적으로 표현하는 데에 사용

: 딥러닝 모델의 깊이가 깊어지면 기울기가 사라지는 '기울기 소멸 문제' 발생

f(x) = 1 / (1+e^(-x))

- 하이퍼볼릭 탄젠트 함수

: 선형 함수의 결과를 0~1 사이에서 비선형 형태로 변형

- 렐루 함수

: 입력이 음수일 때는 0 출력, 양수일 때는 x 출력

: 경사 하강법에 영향 주지 X 학습 속도가 빠르고, 기울기 소멸 문제 발생 X

: 일반적으로 은닉층에서 사용됨.

: 하이퍼볼릭 탄젠트 함수 대비 학습 속도가 6배 빠름.

: 음수 값을 입력받으면 항상 0 값 출력 (학습 능력 감소) -> 해결 방법: 리키 렐루 함수

- 리키 렐루 함수

: 입력 값이 음수이면 0이 아닌 0.001처럼 매우 작은 수를 반환

: 입력 값이 수렴하는 구간 제거 (렐루 함수의 문제점 해결)

- 소프트맥스 함수

: 입력 값을 0~1 사이에 출력되도록 정규화하여 출력 값들의 총합=1이 되도록 함.

: 딥러닝에서 출력 노드의 활성화 함수로 많이 사용됨.

yk = exp(ak) / 시그마(i=1~n)exp(ai)

(n: 출력층의 뉴런 개수, yk: k번째 출력)

(분자: 입력 신호 ak의 지수 함수, 분모: 모든 입력 신호의 지수 함수 합)

손실 함수: 가중치 학습을 위해 출력 함수의 결과와 실제 값 간의 오차를 측정하는 함수

```python
import torch
```

```python
# 렐루 함수, 소프트맥스 함수를 파이토치에서 구현하는 코드

class Net(torch.nn.Module):
  def __init__(self, n_feature, n_hidden, n_output):
    super(Net, self).__init__()
    self.hidden=torch.nn.Linear(n_feature, n_hidden) # 은닉층
    self.relu=torch.nn.Relu(inplace=True)
    self.out=torch.nn.Linear(n_hidden, n_output) # 출력층
    self.softmax=torch.nn.Softmax(dim=n_output)
  def forward(self, x):
    x=self.hidden(x)
    x=self.relu(x) # 은닉층을 위한 렐루 활성화 함수
    x=self.out(x)
    x=self.softmax(x) # 출력층을 위한 소프트맥스 활성화 함수
    return x
```

- 손실 함수: 학습을 통해 얻은 데이터의 추정치가 실제 데이터와 얼마나 차이가 나는지 평가하는 지표

: 손실 함수 값이 클수록 많이 틀렸다는 의미이고, 0에 가까울수록 완벽한 추정을 의미

: 대표 손실 함수 - 평균 제곱 오차, 크로스 엔트로피 오차

- 경사 하강법: 학습률, 손실 함수의 순간 기울기를 이용하여 가중치를 업데이트하는 방법

: 미분의 기울기를 이용하여 오차를 비교하고 최소화하는 방향으로 이동시키는 방법

- 평균 제곱 오차: 실제 값, 예측 값의 차이를 제곱하여 평균을 낸 것

이전 시점의 정보를 현재 시점의 계산에 반영하여 순서가 있는 데이터를 학습하는 신경망

- 특징

: 이전 시점의 정보가 현재 결과에 영향을 줌.

: 시간적 의존성을 학습할 수 있음.

: 길이가 가변적인 순차 데이터를 처리할 수 있음.

: 긴 시퀀스를 학습할 때 기울기 소멸 문제가 발생할 수 있음.

- 문제 해결 방법

: LSTM: 메모리 셀과 게이트 구조를 이용하여 장기 의존성 학습을 개선한 RNN의 일종

- 활용 분야

: 자연어 처리(NLP), 언어 모델링 및 텍스트 생성, 기계 번역, 음성 인식, 이미지 캡션 생성, 시계열 데이터 분석

# 4.3.4. 제한된 볼츠만 머신 (RBM - Restricted Boltzmann Machine)

가시층과 은닉층으로 구성된 확률 기반 신경망

- 구조적 특징

: 가시층과 은닉층 사이에만 연결이 존재함.

: 가시층 내부 및 은닉층 내부에는 연결이 없음.

- 주요 특징 및 활용

: 차원 감소, 특성 학습, 협업 필터링 등에 활용됨.

: 비지도 학습을 통해 데이터의 특징을 학습할 수 있음.

: 신경망의 사전 학습에 활용 가능함.

: 심층 신뢰 신경망의 기본 구성 요소로 사용됨.

# 4.3.4. 심층 신뢰 신경망 (DBN - Deep Belief Network)

여러 개의 RBM을 층층이 쌓아 구성한 심층 확률 신경망

- 특징

: 비지도 학습을 통해 레이블이 없는 데이터에서도 특징을 학습할 수 있음.

: 아래층에서 위층으로 갈수록 추상적인 특징을 학습함.

: 계층별 사전 학습을 통해 신경망의 가중치를 초기화할 수 있음.

: 학습한 가중치를 다층 퍼셉트론의 초기 가중치로 활용할 수 있음.

- 학습 절차

1. 가시층과 첫 번째 은닉층 사이의 RBM을 학습함.

2. 첫 번째 RBM의 학습 결과를 이용하여 다음 RBM을 학습함.

3. 원하는 층 개수만큼 RBM을 순차적으로 쌓음.

4. 전체 DBM을 완성하고 필요에 따라 추가 학습을 수행함.
