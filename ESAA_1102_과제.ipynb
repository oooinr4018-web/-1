# 5.1. 합성곱 신경망

오차 정보가 신경망의 모든 노드(출력층->은닉층->입력층)로 전송

이 과정 계산 오래 걸림. -> 해결 방법: 합성곱 신경망

- 합성곱 신경망: 이미지 전체를 한 번에 계산 X, 국소적 부분 계산, 이미지의 세밀한 부분까지 분석

# 5.1.1. 합성곱층의 필요성

펼쳐서(flattering) 각 픽셀에 가중치를 곱하여 은닉층으로 전달

(데이터의 공간적 구조 무시)

-> 합성곱층은 이를 방지.



# 5.1.2. 합성곱 신경망 구조

합성곱 신경망: 음성 인식, 이미지/영상 인식에서 주로 사용되는 신경망

: 다차원 배열 데이터 처리

(컬러 이미지같은 다차원 배열 처리에 특화)

: 계층 5개(입력층, 합성곱층, 풀링층, 완전연결층, 출력층)로 구성

: 과정

(1) 합성곱층, 풀링층을 거치며 입력 이미지의 주요 특성 벡터(feature vector) 추출

(2) 완전연결층을 거치며 1차원 벡터로 변환

(3) 출력층에서 활성화 함수인 소프트맥스 함수 사용

(4) 최종 결과 출력



- 입력층

: 입력 이미지 데이터가 최초로 거치는 계층

: 이미지는 3차원 데이터

(이미지가 그레이스케일(gray scale)이면 1 값, 컬러(RGB)면 3 값)

(ex - 높이 4, 너비 4, 채널 RGB -> shape: (4,4,3)

- 합성곱층

: 입력 데이터에서 특성을 추출

: 특성 추출 과정

커널/필터: 이미지의 모든 영역 훑으며 특성 추출 -> 결과물: 특성 맵

(커널 3x3, 5x5 크기로 적용)

(스트라이드라는 지정된 간격에 따른 순차적 이동)

ex - 그레이스케일 이미지의 합성곱

스트라이드가 1일 때 이동 과정

(이미지 크기 (6,6,1), 커널/필터(3x3) 가 스트라이드 1 간격으로 이동

1단계) 입력 이미지, 필터(3x3)의 대응되는 숫자끼리 곱한 후 모두 더하기 -> 출력 값

2단계) 입력 이미지에 필터가 1만큼 이동 (1칸씩 오른쪽으로 이동 후, 1칸씩 아래쪽으로 이동) -> 계속하여 출력 값 내놓음.

((원본 (6,6,1) -> 특성 맵 (4,4,1)))

: 컬러 이미지

(1) 필터 채널이 3 (필터 개수는 1개)

(2) RGB 각각에 서로 다른 가중치로 합성곱을 적용하여 결과 더함.

ex - 컬러 이미지의 합성곱

입력 이미지(3x3x3) 3개, 필터(2x2x3) 3개 (채널 수 '3'으로 동일)

각각 입력 이미지와 필터의 합성곱 연산을 수행한 후 각각의 자리 값들을 더하기 -> 특성 맵 값

: 입력 데이터 (W1(가로)xW2(세로)x D1(채널(깊이)))

+하이퍼파라미터 (K(필터 개수), F(필터 크기), S(스트라이드), P(패딩))

+출력 데이터 (W2=(W1-F+2P)/S+1, H2=(H1-F+2P)/S+1, D2=K)

- 풀링층

: 특성 맵의 차원 다운 샘플링 -> 연산량 감소, 주요 특성 벡터 추출

다운 샘플링: 이미지 축소

: 풀링 연산

(1) 최대 풀링: 대상 영역에서 최댓값 추출

(대부분의 합성곱 신경망에서 사용 - 평균 풀링은 각 커널 값을 평균화시켜 중요한 가중치를 갖는 값의 특성이 희미해질 가능성)

(2) 평균 풀링: 대상 영역에서 평균 반환

: 입력 데이터(W1xH1xD1)

+하이퍼파라미터(F(필터 크기), S(스트라이드))

+출력 데이터(W2=(W1-F)/S+1, H2=(H1-F)/S+1, D2=D1)

- 완전연결층

: 합성곱층, 풀링층 거쳐 차원이 축소된 특성 맵 -> 완전연결층으로 전달

(이미지가 3차원 벡터 -> 1차원 벡터로 flatten)

- 출력층

: 마지막 출력층의 소프트맥스 함수 사용하여 이미지가 각 레이블에 속할 확률 값 출력 -> 최종 값: 가장 높은 확률 값을 갖는 레이블

: 입력받은 값을 0~1 사이의 값으로 출력














# 5.1.3. 1D, 2D, 3D 합성곱

합성곱은 이동하는 방향의 수, 출력 형태에 따라 1D, 2D, 3D로 분류

- 1D 합성곱

: 필터가 시간을 축으로 좌우로만 이동할 수 있는 합성곱

: 입력(W), 필터(k) -> 출력: W

ex - 입력 [1, 1, 1, 1, 1], 필터 [0.25, 0.5, 0.25] -> 출력 [1, 1, 1]

: 출력 형태 - 1D 배열

: 그래프 곡선 완화

- 2D 합성곱

: 필터가 방향 2개로 움직이는 형태

: 입력(W,H), 필터(k,k) -> 출력(W,H)

: 출력 형태 - 2D 행렬

- 3D 합성곱

: 입력(W,H,L), 필터(k,k,d) -> 출력 (W,H,L)

(d < L 유지해야함.)

: 출력 형태 - 3D 형태




- 3D 입력을 갖는 2D 합성곱

: 입력이 3D 형태임에도 출력 형태가 3D가 아닌 2D 행렬

(필터 길이(L)가 입력 채널의 길이(L)과 같아야 함.)

: 입력(W,H,L)에 필터(k,k,L) 적용 -> 출력 (W,H)

: 대표적 사례 - LeNet-5, VGG

- 1x1 합성곱

: 3D 형태로 입력

: 입력(W,H,L) + 필터(1,1,L) -> 출력(W,H)

: 채널 수 조정 -> 연산량 감소

: 대표적 사례 - GoogLeNet
