# 5. Details of Learning

(1) 학습 방법: Stochastic Gradient Descent (SGD)

- Batch size(집단 크기): 128

- Momentum(추진력): 0.9

- Weight decay(가중치 감소): 0.0005

- 초기 learning rate: 0.01

- Validation error(검증 오류)가 개선되지 않으면, learning rate를 1/10로 감소

- Weight:

평균 0

standard deviation 0.01인 Gaussian distribution으로 초기화

- 일부 layer의 bias를 1로 초기화하여 초기 ReLU 학습 가속

- 약 90 cycles(epochs)동안 학습

- GPU: NVIDIA GTX 580 3GB x 2

- 전체 학습 시간: 약 5~6일

# 6.1 Qualitative Evaluations

CNN이 학습한 convolution kernel 분석

- 첫 번째 convolution layer에서

방향성(edge/orientation) feature, 주파수 관련 feature, 색상 관련 feature, 다양한 형태의 visual pattern 등을 학습

- 두 GPU가 서로 다른 feature에 특화되는 현상 관찰

GPU 1: 색상에 덜 의존하는 feature

GPU 2: 색상에 특화된 featire

- 마지막 4096-dimensional feature를 이용해 이미지 간 similarity 분석

- pixel 수준에서는 다른 이미지도 CNN의 feature space에서는 유사하게 표현될 수 있음

- CNN이 단순한 pixel pattern뿐 아니라 높은 수준의 visual representation 학습






# 7. Discussion

(1) 결론

- CNN의 depth가 성능에 중요

- convolutional layer 하나를 제거하면 성능 저하

*중간 later를 제거하면 Top-1 accuracy 성능이 약 2% 감소

-> parameter 수를 늘리는 것뿐만 아니라 여러 층을 통한 hierarchical feature learning이 중요함

- 별도의 unsupervised pre-training 없이 supervised learning만으로 높은 성능 달성

- 네트워크를 더 크게 만들고 더 오래 학습할수록 성능이 향상되는 경향
