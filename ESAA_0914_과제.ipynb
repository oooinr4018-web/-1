```python
import nltk

from nltk.tokenize import word_tokenize

from nltk.stem import WordNetLemmatizer


# 필수 NLTK 데이터 다운로드(필요시)

nltk.download('punkt')

nltk.download('punkt_tab')

nltk.download('wordnet')


def LemNormalize(text):
  lemmatizer = WordNetLemmatizer()
  return [lemmatizer.lemmatize(token) for token in word_tokenize(text)]
```

```python
import pandas as pd
import glob,os
import warnings
warnings.filterwarnings('ignore')
pd.set_option('display.max_colwidth', 700)

# 다음은 저자의 컴퓨터에서 압축 파일을 풀어놓은 디렉터리이니, 각자 디렉터리를 다시 설정합니다.
path = '/content/opinosis/OpinosisDataset1.0/topics'
# path로 지정한 디렉터리 밑에 있는 .data 파일들의 파일명을 리스트로 취합
all_files=glob.glob(os.path.join(path, "*.data"))
filename_list=[]
opinion_text=[]

# 개별 파일들의 파일명은 filename_list 리스트로 취합,
# 개별 파일들의 파일 내용은 DatFrame 로딩 후 다시 ㄴtring으로 변환하여 opinion_text 리스트로 취합
for file_ in all_files:
  # 개별 파일을 읽어서 DataFrame으로 생성
  df=pd.read_table(file_, index_col=None, header=0, encoding='latin1')
  # 절대경로로 주어진 파일명을 가공, Linux에서 수행 시에는 아래 \\를 / 변경.
  # 맨 마지막 .data 확장자도 제거
  filename_=file_.split('/')[-1]
  filename=filename_.split('.')[0]

  # 파일명 리스트와 파일 내용 리스트에 파일명과 파일 내용을 추가.
  filename_list.append(filename)
  opinion_text.append(df.to_string())

# 파일명 리스트와 파일 내용 리스트를 DataFrame으로 생성
document_df=pd.DataFrame({'filename':filename_list, 'opinion_text':opinion_text})
document_df.head()
```

- 각 파일 이름(filename) 자체만으로 의견의 텍스트가 어떠한 제품/서비스에 대한 리뷰인지 확인 가능

- 문서를 TF-IDF 형태로 피처 벡터화

- TfidfVectorizer: tokenizer 인자에 커스텀 어근 변환 함수를 적용하여 어근 변환 수행 (Lemmatization과 같은 어근 변환 직접 지원 X)

- ngram=(1,2)

- 피처 개수 제한: min_df/max_df 범위 설정

- 개별 문서 텍스트에 대한 TF-IDF 변환된 피처 벡터화된 행렬 구하기: TfidfVectorizer의 fit_transform() 인자로 document_df DataFrame의 opinion_Text 칼럼 입력

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vect=TfidfVectorizer(tokenizer=LemNormalize, stop_words='english', \
                           ngram_range=(1,2), min_df=0.05, max_df=0.85 )
# opinion_Text 칼럼값으로 feature vectorization 수행
feature_vect=tfidf_vect.fit_transform(document_df['opinion_text'])
```

-  TF-IDF 변환 피처 벡터화 행렬 데이터의 군집화 수행 -> 어떤 문서끼리 군집되늕 확인

- 군집화 기법: K-평균

- 5개의 중심(Centroid) 기반 군집화 확인

- 최대 반복 횟수 max_iter=10000

- KMeans 수행

- 군집 Label 값, 중심별로 할당된 데이터 세트의 좌표 값 구하기

# 군집별 핵심 단어 추출하기

- 각 군집(Cluster)에 속한 문서는 핵심 단어를 주축으로 군집화

-KMeans 객체는 각 군집을 구성하는 단어 피처가 군집의 중심(Centroid)을 기준으로 얼마나 가깝게 위치해 있는지 clusters_centers_ 속성으로 제공

clusters_centers_: 배열 값 (행- 개별 군집, 열- 개별 피처)

clusters_centers[0,1]: 0번 군집의 두 번째 피처의 위치 값

```python
# 앞 예제에서 군집 3개로 생성한 KMenas 객체 km_cluster에서 cluster_centers_ 속성값, 값 확ㅇ;ㄴ

cluster_centers=km_cluster.cluster_centers_
print('cluster_centers shale:', cluster_centers.shape)
print(cluster_centers)
```

- cluster_centers_: (3,6155) 배열 (3개의 군집 4611개의 피처)

- 각 행의 배열 값: 각 군집 내의 4611개의 피처의 위치가 개별 중심과 얼마나 가까운지 상대 값으로 나타낸 값

(0~1까지의 값, 1에 가까울수록 중심과 가까운 값)

- cluster_centers_ 속성값으로 각 군집별 핵심 단어 찾아보기

(cluster_centers_ 속성: 넘파이의 ndarray)

- ndararray의 argsort()[:,::-1]: cluster_centers 배열 내 값이 큰 순으로 정렬된 위치 인덱스 값 반환 (핵심 단어 피처의 이름 출력)

- get_cluster_details() 함수

1. ㅊluster_centers_ 뱅려 내에서 가장 값이 큰 데이터의 위치 인덱스 추출

2. 해당 인덱스를 이용하여 핵심 단어 이름, 상대 위치 값 추출 (cluster_details라는 Dic 객체 변수에 기록, 반환)



```python
def get_cluster_details(cluster_model, cluster_data, feature_names,
                        clusters_num, top_n_features=10):

    cluster_details = {}

    # 각 군집 중심에서 중요도가 높은 feature 순서
    centroid_feature_ordered_ind = \
        cluster_model.cluster_centers_.argsort()[:, ::-1]

    for cluster_num in range(clusters_num):

        cluster_details[cluster_num] = {}

        # 상위 feature들의 index
        top_feature_indexes = \
            centroid_feature_ordered_ind[cluster_num, :top_n_features]

        # 상위 feature 이름
        top_features = [
            feature_names[ind] for ind in top_feature_indexes
        ]

        # 상위 feature 값
        top_feature_values = \
            cluster_model.cluster_centers_[
                cluster_num, top_feature_indexes
            ].tolist()

        # 군집별 정보 저장
        cluster_details[cluster_num]['cluster'] = cluster_num
        cluster_details[cluster_num]['top_features'] = top_features
        cluster_details[cluster_num]['top_feature_values'] = top_feature_values

        # 해당 군집에 포함된 파일명
        filenames = cluster_data[
            cluster_data['cluster_label'] == cluster_num
        ]['filename']

        filenames = filenames.values.tolist()

        cluster_details[cluster_num]['filenames'] = filenames

    return cluster_details
```

- get_cluster_details()

: dictionary를 원소로 가지는 리스트인 cluster_details 반환

: cluster_details: 개별 군집번호, 핵심 단어, 핵심단어 중심 위치 상댓값, 파일명 속성 값 정보 보유

```python
# print_cluster_details() 함수 생성

def print_cluster_details(cluster_details):
  for cluster_num, cluster_detail in cluster_details.items():
    print('####### Cluster {0}'.format(cluster_num))
    print('Top features:', cluster_detail['top_features'])
    print('Reviews 파일명 :', cluster_detail['filenames'][:7])
    print('=============================================')
```

```python
# get_cluster_details(), print_cluster_details() 함수 호출
# get_cluster_detail() 인자: KMeans 군집화 객체, 파일명 추출을 위한 document_df DataFrame, 핵심 단어 추출을 위한 피처명 리스트, 전체 군집 개수, 핵심 단어 추출 개수
# TF-IDF 변환된 tfidf_vect 객체에서 get_feature_names()로 피처명 리스트 추출

feature_names=tfidf_vect.get_feature_names_out()

cluster_details=get_cluster_details(cluster_model=km_cluster, cluster_data=document_df,
                                    feature_names=feature_names, clusters_num=3, top_n_features=10 )
print_cluster_details(cluster_details)
```
