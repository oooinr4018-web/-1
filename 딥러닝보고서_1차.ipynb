# 3.1 ReLU Nonlinearity

(1) 기존 activation

- tanh

- sigmoid

(2) 모델 설명

- AlexNet에서는 ReLU(Rectified Linear Unit) 사용

ReLU: f(x) = max(0,x)

(3) 장점

- gradient가 빠르게 전달되어 학습 속도가 빠름

- tank를 사용하는 네트워크보다 훨씬 빠르게 학습 가능

- CIFAR-10 실험에서,

ReLU network가 tank network보다 약 6배 빠르게 25% training error에 도달

- 대규모 CNN을 실제로 학습하기 위해 중요한 요소

#3.2 Training on Multiple GPUs

(1) 기존 모델의 한계

- 네트워크가 너무 커서 GPU 하나의 메모리로 처리하기 어려움

-> 2개의 GPU에 네트워크를 분산하여 학습

(2) 모델 설명

- GPU끼리 모든 layer에서 통신하지 않고, 특정 layer에서만 통신

- GPU 간 communication cost를 줄이면서 큰 모델 학습 가능

- 결과:

TOP-1 error 1.7% 감소

Top-5 error 1.2% 감소




# 3.3 Local Response Normalization

(1) 모델 설명

- ReLU 이후 특정 neuron의 activation이 지나치게 커지는 것 조절

- 인접한 feature map의 activation을 이용해 normalization

- 생물학적 뉴런의 lateral inhibition과 유사한 개념

- 일반화 성능 향상

- 결과:

Top-1 error 1.4% 감소

Top-5 error 1.2% 감소

# 4. Reducing Overfitting

(1) 문제점

- AlexNet은 약 6천만 개의 parameter를 가지고 있어 overfitting의 가능성

- 1.2 million training images만으로 충분하지 않을 가능성

(2) 해결 방안

- Data Augmentation

- Dropout



# 4.1 Data Augmentation

training data를 인위적으로 증가시켜 overfitting 감소

(1) 방법 1: Random Crop + Horizontal Reflection

- 256 x 256 이미지에서 224 x 224 patch 추출

- 좌우 반전도 사용

- training set을 이론적으로 2048배 증가

- 테스트시:

4개 corner + 1개 center crop

각각 horizontal reflection

총 10개 patch의 prediction을 평균

(2) 방법 2: PCA-based RGB intensity augmentation

- RGB pixel에 PCA 기반 변화 추가

- 조명 및 색 변화에 대한 robustness(억셈) 증가

- Top-1 error 1% 이상 감소


# 4.2 Dropout

Hidden neuron의 출력을 50% 확률로 0으로 설정

(1) 효과

- 매번 다른 neuron이 제거되므로 다양한 네트워크 구조 학습

- 특정 neuron에 지나치게 의존하는 co-adaptation 방지

- 더 robust한 feature를 학습하도록 유도

- 첫 번째와 두 번째 fully-connected layer에 적용

- dropout을 사용하지 않을 시 상당한 overfitting 발생
